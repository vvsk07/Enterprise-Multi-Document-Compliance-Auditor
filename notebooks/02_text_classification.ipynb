This notebook demonstrates the classical NLP pipeline built using:

- **spaCy (`en_core_web_sm`)** for Named Entity Recognition (NER)
- **Custom Domain Heuristics** for contract IDs, deadlines, and policy numbers

## 1. Train & Evaluate Document Classifier

```python
metrics = classification_service.train('../data/training/document_classification.csv')
print('Classification Evaluation Metrics:')
for k, v in metrics.items():
    print(f' - {k:<20}: {v}')
```

## 2. Test Document Classification Prediction

```python
sample_contract = "This Master Services Agreement is entered into by Acme Corp and Supplier for IT consulting services."
predicted_type, confidence = classification_service.predict(sample_contract)
print(f'Predicted Category : {predicted_type}')
print(f'Confidence Score   : {confidence:.4f}')
```

## 3. Extract Named Entities & Normalize

```python
sample_text = "Acme Global Inc entered agreement AGR-2026-991 with Vertex LLC on January 15, 2026 for $50,000 USD."
entities = ner_service.extract_entities(sample_text)
for e in entities:
    norm_text = entity_normalizer.normalize_entity_text(e['text'], e['label'])
    print(f"{e['label']:<12} | Raw: {e['text']:<25} | Normalized: {norm_text}")
```
